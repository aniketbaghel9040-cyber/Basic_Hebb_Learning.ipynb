```python
import numpy as np

# Input data
X = np.array([
    [1, 1],
    [1, -1],
    [-1, 1],
    [-1, -1]
])

# Target output
Y = np.array([1, -1, -1, 1])

# Initialize weights
w = np.zeros(2)
learning_rate = 1

print("Initial Weights:", w)

# Hebb Learning
for i in range(len(X)):
    w = w + learning_rate * X[i] * Y[i]
    print(f"After Sample {i+1}: {w}")

print("\nFinal Weights:", w)
```

```python
import numpy as np

X = np.array([
    [1, 1],
    [1, -1],
    [-1, 1],
    [-1, -1]
])

Y = np.array([1, -1, -1, 1])

w = np.zeros(2)

# Training
for i in range(len(X)):
    w = w + X[i] * Y[i]

print("Final Weights:", w)

print("\nPrediction")
for x in X:
    net = np.dot(x, w)
    if net >= 0:
        output = 1
    else:
        output = -1

    print("Input:", x, "Output:", output)
```
