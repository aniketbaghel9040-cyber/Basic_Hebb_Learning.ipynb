```python
# BIOENERGY POTENTIAL CALCULATOR
# Researchṇ Topic: Bioenergy

import pandas as pd
import matplotlib.pyplot as plt

# Biomass data
data = {
    "Biomass Type": [
        "Agricultural Waste",
        "Animal Waste",
        "Food Waste",
        "Wood Waste",
        "Organic Waste"
    ],

    # Amount of waste in kg
    "Waste (kg)": [1000, 500, 800, 600, 700],

    # Approximate energy potential in kWh per kg
    "Energy per kg (kWh)": [4.0, 2.5, 3.0, 4.5, 2.0]
}

df = pd.DataFrame(data)

# Calculate total energy
df["Energy Potential (kWh)"] = (
    df["Waste (kg)"] * df["Energy per kg (kWh)"]
)

# Display dataset
print("BIOENERGY DATA")
display(df)

# Total energy
total_energy = df["Energy Potential (kWh)"].sum()

print("\nTotal Bioenergy Potential:")
print(total_energy, "kWh")

# Graph
plt.figure(figsize=(10, 5))

plt.bar(
    df["Biomass Type"],
    df["Energy Potential (kWh)"]
)

plt.xlabel("Biomass Type")
plt.ylabel("Energy Potential (kWh)")
plt.title("Bioenergy Potential from Different Biomass Sources")
plt.xticks(rotation=30)
plt.show()
```

```python
# Empty values ko fill karna
for col in ["Title", "Abstract", "Author Keywords", "Index Keywords"]:
    df[col] = df[col].fillna("").astype(str)

# Saare important text columns combine karna
df["Text"] = (
    df["Title"] + " " +
    df["Abstract"] + " " +
    df["Author Keywords"] + " " +
    df["Index Keywords"]
)

# Bioenergy-related papers filter karna
bio_df = df[
    df["Text"].str.contains(
        r"\bbioenergy\b|\bbiomass\b|\bbiofuel\b|\bbiogas\b",
        case=False,
        regex=True,
        na=False
    )
].copy()

print("Total Bioenergy Papers:", len(bio_df))
```

```python
import re
import nltk

nltk.download("stopwords")

from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

def clean_text(text):
    text = text.lower()

    # Special characters remove
    text = re.sub(r"[^a-zA-Z\s]", " ", text)

    # Extra spaces remove
    text = re.sub(r"\s+", " ", text)

    # Stopwords remove
    words = [
        word for word in text.split()
        if word not in stop_words and len(word) > 2
    ]

    return " ".join(words)

bio_df["Clean_Text"] = bio_df["Text"].apply(clean_text)

display(bio_df[["Title", "Clean_Text"]].head())
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(
    max_df=0.95,
    min_df=2,
    max_features=5000
)

X = vectorizer.fit_transform(bio_df["Clean_Text"])

print("Document-Term Matrix:", X.shape)
```

```python
from sklearn.decomposition import LatentDirichletAllocation

# Number of topics
num_topics = 5

lda = LatentDirichletAllocation(
    n_components=num_topics,
    random_state=42,
    learning_method="batch"
)

lda.fit(X)

print("LDA Model trained successfully!")
```

```python
feature_names = vectorizer.get_feature_names_out()

def display_topics(model, feature_names, num_words=10):

    for topic_idx, topic in enumerate(model.components_):

        top_words = [
            feature_names[i]
            for i in topic.argsort()[:-num_words - 1:-1]
        ]

        print(
            f"Topic {topic_idx + 1}: "
            + ", ".join(top_words)
        )

display_topics(
    lda,
    feature_names,
    num_words=10
)
```

```python
topic_distribution = lda.transform(X)

bio_df["Dominant_Topic"] = topic_distribution.argmax(axis=1) + 1

display(
    bio_df[
        ["Title", "Dominant_Topic"]
    ].head(20)
)
```

```python
topic_counts = bio_df["Dominant_Topic"].value_counts().sort_index()

print("Number of papers in each topic:")
display(topic_counts)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9,5))

plt.bar(
    topic_counts.index.astype(str),
    topic_counts.values
)

plt.xlabel("Topic Number")
plt.ylabel("Number of Research Papers")
plt.title("Bioenergy Research Topic Distribution")

plt.show()
```

```python
from wordcloud import WordCloud

for topic_idx, topic in enumerate(lda.components_):

    top_words = {
        feature_names[i]: topic[i]
        for i in topic.argsort()[-30:]
    }

    wc = WordCloud(
        width=900,
        height=500,
        background_color="white"
    ).generate_from_frequencies(top_words)

    plt.figure(figsize=(10,5))
    plt.imshow(wc)
    plt.axis("off")
    plt.title(f"Bioenergy Topic {topic_idx + 1}")

    plt.show()
```

```python
# ==========================================
# BIOENERGY - COHERENCE GRAPH
# ==========================================

!pip install -q gensim openpyxl

import pandas as pd
import re
import matplotlib.pyplot as plt
from gensim import corpora
from gensim.models import LdaModel, CoherenceModel

# Load dataset
file_path = "/content/6.13.xlsx"
df = pd.read_excel(file_path)

# Combine text columns
text_columns = [
    "Title",
    "Abstract",
    "Author Keywords",
    "Index Keywords"
]

df["Text"] = df[text_columns].fillna("").astype(str).agg(" ".join, axis=1)

# Select Bioenergy-related papers
bioenergy_df = df[
    df["Text"].str.contains(
        r"\bbioenergy\b|\bbiomass\b|\bbiofuel\b|\bbiogas\b",
        case=False,
        regex=True,
        na=False
    )
].copy()

print("Bioenergy papers:", len(bioenergy_df))


# ------------------------------------------
# Text preprocessing
# ------------------------------------------

stop_words = {
    "the", "and", "for", "with", "from", "that",
    "this", "are", "was", "were", "have", "has",
    "using", "used", "use", "study", "based",
    "research", "paper", "result", "results",
    "method", "also", "can", "may", "into"
}

def preprocess(text):
    text = text.lower()
    text = re.sub(r"[^a-z\s]", " ", text)

    words = text.split()

    words = [
        word for word in words
        if len(word) > 2 and word not in stop_words
    ]

    return words


processed_texts = [
    preprocess(text)
    for text in bioenergy_df["Text"]
]

# Remove empty documents
processed_texts = [
    text for text in processed_texts
    if len(text) > 0
]


# ------------------------------------------
# Dictionary and Corpus
# ------------------------------------------

dictionary = corpora.Dictionary(processed_texts)

dictionary.filter_extremes(
    no_below=2,
    no_above=0.95
)

corpus = [
    dictionary.doc2bow(text)
    for text in processed_texts
]


# ------------------------------------------
# Coherence calculation
# ------------------------------------------

topic_numbers = range(2, 9)

coherence_scores = []

for num_topics in topic_numbers:

    lda_model = LdaModel(
        corpus=corpus,
        id2word=dictionary,
        num_topics=num_topics,
        random_state=42,
        passes=5,
        iterations=50
    )

    coherence_model = CoherenceModel(
        model=lda_model,
        texts=processed_texts,
        dictionary=dictionary,
        coherence="c_v"
    )

    score = coherence_model.get_coherence()

    coherence_scores.append(score)

    print(
        f"Topics = {num_topics}, "
        f"Coherence = {score:.4f}"
    )


# ------------------------------------------
# Coherence Graph
# ------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    list(topic_numbers),
    coherence_scores,
    marker="o"
)

plt.xlabel("Number of Topics")
plt.ylabel("Coherence Score")
plt.title("Coherence Score vs Number of Topics - Bioenergy")

plt.xticks(list(topic_numbers))
plt.grid(True)

plt.show()


# ------------------------------------------
# Best Topic Number
# ------------------------------------------

best_index = coherence_scores.index(max(coherence_scores))

best_topic = list(topic_numbers)[best_index]
best_score = coherence_scores[best_index]

print("\nBest Number of Topics:", best_topic)
print("Best Coherence Score:", round(best_score, 4))
```
